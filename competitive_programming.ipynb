```python
#duplicate identify

def contains_duplicate(order_ids):
    seen = set()
    for transaction_id in order_ids:
        if transaction_id in seen:
            return True
        seen.add(transaction_id)
    return False
order_ids = [101, 204, 305, 101, 509]

if contains_duplicate(order_ids):
    print("True")
else:
    print("False")
```

```python
# two pointer traversal
def move_zeros(logs):
    pos = 0
    for i in range(len(logs)):
        if logs[i] != 0:
            logs[pos] = logs[i]
            pos += 1
    while pos < len(logs):
        logs[pos] = 0
        pos += 1
logs = [0,200,0,400,500,0,301]
move_zeros(logs)
print(logs)
```
