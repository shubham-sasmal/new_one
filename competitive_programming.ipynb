```python


def contains_duplicate(order_ids):
    seen = set()
    for transaction_id in order_ids:
        if transaction_id in seen:
            return True
        seen.add(transaction_id)
    return False
order_ids = [101, 204, 305, 101, 509]

if contains_duplicate(order_ids):
    print("True")
else:
    print("False")
```

```python
logs = [0, 200, 0, 400, 500, 0, 301]

w = 0
r = 0

while r < len(logs):
    if logs[r] == 0:
        r += 1
    else:
        logs[w] = logs[r]
        w += 1
        r += 1

# Fill remaining positions with 0
while w < len(logs):
    logs[w] = 0
    w += 1

print(logs)
```

```python
# Coin Change using Greedy and Dynamic Programming

def greedy(coins, value):
    coins.sort(reverse=True)
    selected = []
    for coin in coins:
        while value >= coin:
            value -= coin
            selected.append(coin)
    print("Greedy Function")
    print("Total required coins =", len(selected))
    print("Coins selected =", selected)
def dynamic(coins, value):
    dp = [float('inf')] * (value + 1)
    choice = [-1] * (value + 1)
    dp[0] = 0
    for amount in range(1, value + 1):
        for coin in coins:
            if coin <= amount and dp[amount - coin] + 1 < dp[amount]:
                dp[amount] = dp[amount - coin] + 1
                choice[amount] = coin
    selected = []
    temp = value
    while temp > 0:
        selected.append(choice[temp])
        temp -= choice[temp]

    print("\nDynamic Function")
    print("Total required coins =", dp[value])
    print("Coins selected =", selected)
V = 6
D = [4, 3, 1]
print("Input:")
print("V =", V)
print("D =", D)
greedy(D.copy(), V)
dynamic(D, V)
```
