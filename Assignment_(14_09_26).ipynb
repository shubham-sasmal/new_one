```python
A = {10, 20, 30, 40, 50}
B = {30, 40, 50, 60, 70}
print("Set A:", A)
print("Set B:", B)
print("1. Union:", A | B)
print("2. Intersection:", A & B)
print("3. Difference A - B:", A - B)
print("4. Difference B - A:", B - A)
print("5. Symmetric Difference:", A ^ B)
if 30 in A:
    print("6. 30 exists in Set A")
else:
    print("6. 30 does not exist in Set A")
```

```python
python_students = {"Amit", "Riya", "Rahul", "Sneha", "Ankit"}
java_students = {"Riya", "Rahul", "Priya", "Ankit", "Karan"}
print("Python Students:", python_students)
print("Java Students:", java_students)
print("1. Both Courses:", python_students & java_students)
print("2. Only Python:", python_students - java_students)
print("3. Only Java:", java_students - python_students)
print("4. Either Python or Java (Not Both):", python_students ^ java_students)
print("5. Total Unique Students:", len(python_students | java_students))
```

```python
A = {1, 2, 3, 4, 5, 6}
B = {2, 3, 4}
C = {7, 8}
print("Set A:", A)
print("Set B:", B)
print("Set C:", C)
print("1. Is B a subset of A?", B.issubset(A))
print("2. Is A a superset of B?", A.issuperset(B))
print("3. Is C a subset of A?", C.issubset(A))
print("4. Are A and C disjoint?", A.isdisjoint(C))
print("5. Are B and C disjoint?", B.isdisjoint(C))
```

```python

fruits = {"Apple", "Banana", "Mango"}
print("Original Set:", fruits)
fruits.add("Orange")
print("\nAfter add():", fruits)
fruits.update({"Grapes", "Pineapple"})
print("After update():", fruits)
fruits.remove("Banana")
print("\nAfter remove():", fruits)
fruits.discard("Cherry")
print("After discard():", fruits)
removed = fruits.pop()
print("\nPopped Element:", removed)
print("After pop():", fruits)
fruits.clear()
print("\nAfter clear():", fruits)
```

```python
python = {"Amit", "Riya", "Rahul", "Sneha"}
java = {"Rahul", "Priya", "Karan", "Sneha"}
sql = {"Amit", "Rahul", "Priya", "Ankit"}
print("Python:", python)
print("Java:", java)
print("SQL:", sql)
print("1. All Three Technologies:", python & java & sql)
print("2. Python and Java:", python & java)
print("3. Python but not Java:", python - java)
print("4. At Least One Technology:", python | java | sql)
exactly_two = ((python & java) - sql) | ((python & sql) - java) | ((java & sql) - python)
print("5. Exactly Two Technologies:", exactly_two)
```
