```python
student = {
    "Roll Number": "259",
    "Name": "Shubham Sasmal",
    "Department": "Cyber Security",
    "Semester": "5th",
    "CGPA": 7.75
}

print("Student Information:")
for key, value in student.items():
    print(key, ":", value)
```

```python
student = {
    "name": "Amit",
    "roll": 101,
    "department": "CSE",
    "cgpa": 8.5
}
print("Student Name:", student["name"])
print("CGPA:", student["cgpa"])
print("Department:", student.get("department"))
if "email" in student:
    print("Email key exists.")
else:
    print("Email key does not exist.")
```

```python
employee = {
    "id": 101,
    "name": "Rahul",
    "salary": 45000
}
employee["department"] = "CSE"
employee["salary"] = 50000
employee["experience"] = 5
print("Updated Employee Dictionary:")
for key, value in employee.items():
    print(key, ":", value)
```

```python
marks = {
    "Math": 90,
    "Physics": 85,
    "Chemistry": 88,
    "English": 92,
    "Computer": 95
}
del marks["Chemistry"]
marks.pop("Physics")
marks.popitem()
print("Remaining Dictionary:")
for key, value in marks.items():
    print(key, ":", value)
```

```python
dict1 = {"A": 10, "B": 20, "C": 30}
dict2 = {"D": 40, "E": 50, "F": 60}
merged_dict = dict1 | dict2
print("Merged Dictionary:")
print(merged_dict)
dict3 = {"A": 10, "B": 20, "C": 30}
dict4 = {"B": 200, "C": 300, "D": 40}
merged_common = dict3 | dict4
print("\nMerged Dictionary with Common Keys:")
print(merged_common)
```
