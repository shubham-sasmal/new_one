```python
numbers = [12, 7, 15, 7, 20, 9, 12, 25, 7, 18]
print("List:", numbers)
length = len(numbers)
print("Length:", length)
maximum = numbers[0]
minimum = numbers[0]
for n in numbers:
    if n > maximum:
        maximum = n
    if n < minimum:
        minimum = n
print("Maximum:", maximum)
print("Minimum:", minimum)
total = 0
for n in numbers:
    total += n
print("Sum:", total)
mean = total / length
print("Mean:", mean)
sorted_list = numbers[:]
for i in range(length):
    for j in range(0, length - i - 1):
        if sorted_list[j] > sorted_list[j + 1]:
            sorted_list[j], sorted_list[j + 1] = sorted_list[j + 1], sorted_list[j]
if length % 2 == 0:
    median = (sorted_list[length // 2 - 1] + sorted_list[length // 2]) / 2
else:
    median = sorted_list[length // 2]
print("Median:", median)
mode = numbers[0]
max_count = 0
for i in numbers:
    count = 0
    for j in numbers:
        if i == j:
            count += 1
    if count > max_count:
        max_count = count
        mode = i
print("Mode:", mode)
```

```python
# add, insert, remove
students = ["Rahul", "Amit", "Priya", "Sneha", "Riya"]

print("Original list:", students)
students.append("Karan")
students.insert(2, "Neha")
students.remove("Amit")

print("Final list:", students)
```

```python
# Remove Duplicate Elements from a List

numbers = [1, 2, 3, 2, 4, 5, 1, 6, 3]

unique = []

for n in numbers:
    if n not in unique:
        unique.append(n)

print("Original list:", numbers)
print("List after removing duplicates:", unique)
```

```python
#Reverse a List without Using reverse()

numbers = [10, 20, 30, 40, 50]
reversed_list = []
for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])
print("Original list:", numbers)
print("Reversed list:", reversed_list)
```

```python
# Frequency finding

numbers = [2, 5, 3, 2, 8, 2, 9, 5]

element = int(input("Enter the element to find frequency: "))

count = 0

for n in numbers:
    if n == element:
        count += 1

print("Frequency of", element, "is", count)
```
