```python

my_tuple = (10, 20, 30, 40, 50)
print(my_tuple)
```

```python
# Tuple with different data types
my_tuple = (10, "Hello", 3.14, True)
print(my_tuple)
```

```python
#Create a tuple of numbers and print one item

numbers = (10, 20, 30, 40, 50)
print(numbers[2])
```

```python
#Unpack a tuple into several variables

my_tuple = (10, 20, 30)
a, b, c = my_tuple
print(a)
print(b)
print(c)
```

```python
#Add an item to a tuple

my_tuple = (10, 20, 30)
my_tuple = my_tuple + (40,)
print(my_tuple)
```

```python
#Get the 4th element from the last element of a tuple


my_tuple = (10, 20, 30, 40, 50, 60, 70)
print(my_tuple[-4])
```

```python
# Find repeated items in a tuple

my_tuple = (10, 20, 30, 20, 40, 10, 50, 20)
repeated = []
for item in my_tuple:
    if my_tuple.count(item) > 1 and item not in repeated:
        repeated.append(item)
print("Repeated items:", repeated)
```

```python
#Remove an item from a tuple

my_tuple = (10, 20, 30, 40, 50)
my_list = list(my_tuple)
my_list.remove(30)
my_tuple = tuple(my_list)
print(my_tuple)
```

```python
#Slice a tuple

my_tuple = (10, 20, 30, 40, 50)
print(my_tuple[1:4])
```

```python
#Find the length of a tuple

my_tuple = (10, 20, 30, 40, 50)
print("Length:", len(my_tuple))
```

```python
#Reverse a tuple

my_tuple = (10, 20, 30, 40, 50)
reversed_tuple = my_tuple[::-1]
print(reversed_tuple)
```

```python
#Convert a list of tuples into a dictionary

my_list = [("a", 1), ("b", 2), ("c", 3)]
my_dict = dict(my_list)
print(my_dict)
```
